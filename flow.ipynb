## PHẦN 1: EXPLORATORY DATA ANALYSIS (EDA)

### 1. Sector Analysis (Phân tích Ngành)

- Phân phối các mã theo nhóm ngành.

- Độ tương quan (Correlation) giữa các ngành.

### 2. Stock Characteristics (Đặc điểm Cổ phiếu)

- Phân tích các đặc tính thống kê (Mean, Std, Skewness, Kurtosis).

- Phân tích thanh khoản (Volume, Turnover) và biến động (Volatility).

- Phân tích, phân loại asset universe, xem stock phù hợp với hệ trending hay mean reversion 

## PHẦN 2: TRADING SIGNAL ANALYSIS
Xây dựng và kiểm thử chiến thuật giao dịch cho từng mã cổ phiếu.

### 1. Feature Engineering (Tạo đặc trưng)
- Xây dựng các chỉ báo kỹ thuật (Technical Indicators).
- Biến đổi dữ liệu (Log return, Normalization).

### 2. Signal Generation (Sinh tín hiệu)
- Logic vào lệnh (Entry Rules).
- Logic thoát lệnh (Exit Rules).

### 3. Risk Management (Quản trị rủi ro)
- Stop Loss / Take Profit: Các quy tắc cắt lỗ và chốt lời động/tĩnh.

- Position Sizing: Xác định khối lượng giao dịch cho mỗi lệnh.

### 4. Backtesting (Kiểm thử quá khứ)
- Mô phỏng giao dịch trên dữ liệu lịch sử.


### 5. Analysis (Phân tích hiệu suất)
- Statistical Measurement: Win Rate, Profit Factor, Sharpe Ratio, Drawdown.

- Visualization: Các measurement, Biểu đồ đường cong vốn (Equity Curve), Biểu đồ lệnh (Trade Plot), Drawdown Chart.

## PHẦN 3: PORTFOLIO ANALYSIS

Đề xuất các Alpha Factors : Momentum, Quality, Volatility...

### 2. Factor Conducting& Evaluation (Đánh giá yếu tố)
- Factor Scoring: Tính điểm số (Z-score) cho từng mã dựa trên yếu tố.

- Factor Quality: Kiểm định IC  IC Decay, Neutral- Dollar porforlio,...

### 3. Portfolio Optimization (Tối ưu hóa danh mục)
P-hân bổ trọng số (Weighting): Mean-Variance

- Ràng buộc (Constraints): Giới hạn tỷ trọng ngành, giới hạn turnover.
